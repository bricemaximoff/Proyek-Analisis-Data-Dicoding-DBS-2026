## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Faktor apa saja yang menyebabkan kenaikan angka keterlambatan pengiriman pesanan sebesar 12% pada segmen pelanggan di wilayah São Pauloselama kuartal keempat tahun 2017?

## Exploratory Data Analysis (EDA)

### Pertanyaan 1: "*Faktor apa saja yang menyebabkan kenaikan angka keterlambatan pengiriman pesanan sebesar 12% pada segmen pelanggan di wilayah São Paulo selama kuartal keempat tahun 2017?"*

```python
# Pemfilteran wilayah São Paulo (SP) pada Q4 2017
sp_orders_q4_2017 = orders_df.merge(customers_df, on="customer_id", how="inner")

mask = (
    (sp_orders_q4_2017["customer_state"] == "SP") &
    (sp_orders_q4_2017["order_purchase_timestamp"] >= "2017-10-01") &
    (sp_orders_q4_2017["order_purchase_timestamp"] <= "2017-12-31")
)

sp_orders_q4_2017 = sp_orders_q4_2017.loc[mask].copy()

# Perhitungan status keterlambatan pengiriman
sp_orders_q4_2017.loc[:, "is_late"] = (
    sp_orders_q4_2017["order_delivered_customer_date"] > sp_orders_q4_2017["order_estimated_delivery_date"]
)

# Agregasi untuk melihat persentase keterlambatan
late_rate = sp_orders_q4_2017["is_late"].value_counts(normalize=True) * 100
print("Persentase Keterlambatan Pengiriman di São Paulo (Q4 2017):")
print(late_rate)
```

**Insight:** (Opsional)
- Persentase Keterlambatan: Sekitar 5,58% dari total pesanan di wilayah São Paulo pada kuartal keempat tahun 2017 mengalami keterlambatan pengiriman dari estimasi yang diberikan.

- Dampak Bisnis: Meskipun persentasenya tergolong kecil dibandingkan pesanan yang tepat waktu (94,42%), volume transaksi yang masif di kota besar seperti São Paulo menjelang akhir tahun membuat jumlah absolut pesanan terlambat tetap tinggi, yang berisiko menurunkan tingkat kepercayaan pelanggan.

### Pertanyaan 2: *"Apa penyebab utama terjadinya lonjakan tingkat pembatalan pesanan sebesar 8% pada transaksi yang menggunakan metode pembayaran boleto di kota Rio de Janeiro sepanjang bulan November 2017?"*

```python
# Penggabungan tabel orders, customers, dan order_payments
rj_orders = orders_df.merge(customers_df, on="customer_id", how="inner")
rj_orders = rj_orders.merge(order_payments_df, on="order_id", how="inner")

# Pemfilteran pesanan di kota Rio de Janeiro pada November 2017 dengan metode pembayaran boleto
mask_rj = (
    (rj_orders["customer_city"] == "rio de janeiro") &
    (rj_orders["order_purchase_timestamp"] >= "2017-11-01") &
    (rj_orders["order_purchase_timestamp"] <= "2017-11-30") &
    (rj_orders["payment_type"] == "boleto")
)

rj_nov_2017 = rj_orders.loc[mask_rj].copy()

# Perhitungan distribusi status pesanan
order_status_dist = rj_nov_2017["order_status"].value_counts(normalize=True) * 100
print("Distribusi Status Pesanan Boleto di Rio de Janeiro (November 2017):")
print(order_status_dist)
```

**Insight:** (Opsional)
- Status Transaksi Boleto: Sebagian besar transaksi menggunakan metode pembayaran boleto di Rio de Janeiro pada bulan November 2017 berhasil terkirim sebesar 95,74%.

- Penyebab Pembatalan: Ditemukan sekitar 4,24% total pesanan yang mengalami kendala status non-delivered (terbagi rata pada status canceled, created, invoiced, dan processing masing-masing 1,06%), yang mengindikasikan adanya keterlambatan konfirmasi pembayaran boleto atau pembatalan otomatis oleh sistem akibat batas waktu pembayaran yang kedaluwarsa.

### Pertanyaan 3: *"Apa faktor utama yang memicu penurunan average review score sebesar 0,8 poin pada kategori produk 'eletronicos' dari pelanggan di kota Belo Horizonte selama periode semester pertama tahun 2018?"*

```python
# Penggabungan tabel untuk analisis ulasan produk di Belo Horizonte H1 2018
bh_orders = orders_df.merge(customers_df, on="customer_id", how="inner")
bh_orders = bh_orders.merge(order_items_df, on="order_id", how="inner")
bh_orders = bh_orders.merge(products_df, on="product_id", how="inner")
bh_orders = bh_orders.merge(product_category_translation_df, on="product_category_name", how="inner")
bh_orders = bh_orders.merge(order_reviews_df, on="order_id", how="inner")

# Pemfilteran kategori elektronik/aksesoris komputer di Belo Horizonte pada H1 2018 menggunakan .loc
mask_bh = (
    (bh_orders["customer_city"] == "belo horizonte") &
    (bh_orders["order_purchase_timestamp"] >= "2018-01-01") &
    (bh_orders["order_purchase_timestamp"] <= "2018-06-30") &
    (bh_orders["product_category_name_english"].str.contains("eletr|electro|computer|audio|phone", case=False, na=False))
)

bh_h1_2018 = bh_orders.loc[mask_bh].copy()

# Perhitungan rata-rata skor ulasan dan melihat distribusi skor
print("Rata-rata Review Score:", bh_h1_2018["review_score"].mean())
print("\nDistribusi Review Score:")
print(bh_h1_2018["review_score"].value_counts(sort=False))
```

**Insight:** (Opsional)
- Dominasi Kepuasan Positif: Sebagian besar pelanggan sebenarnya memberikan penilaian sangat puas (bintang 5 sebanyak 78 ulasan), menunjukkan bahwa secara umum produk elektronik yang dijual memiliki daya tarik dan kualitas dasar yang baik.

- Faktor Pemicu Penurunan Skor: Meskipun ulasan positif tinggi, terdapat akumulasi ulasan buruk yang signifikan, terutama ulasan bintang 1 sebanyak 41 ulasan. Angka ulasan bintang 1 yang hampir setara dengan ulasan bintang 4 ini menjadi pemicu utama turunnya rata-rata skor ulasan menjadi 3,60, yang umumnya diakibatkan oleh kendala operasional seperti keterlambatan pengiriman kurir atau kerusakan barang elektronik saat diterima.

## Analisis Lanjutan (Opsional)

### RFM Analysis

RFM Analysis merupakan singkatan dari Recency, Frequency, dan Monetary. RFM Analysis digunakan untuk mengidentifikasi siapa pelanggan terbaik berdasarkan tiga parameter utama:
- Recency: Berapa hari sejak terakhir kali pelanggan melakukan transaksi.
- Frequency: Seberapa sering pelanggan melakukan transaksi pembelian.
- Monetary: Berapa total uang/revenue yang dihabiskan oleh pelanggan.

```python
# 1. Penggabungan tabel orders dan order_items terlebih dahulu untuk mendapatkan total harga per order
orders_items_merged = orders_df.merge(order_items_df, on="order_id", how="inner")

# 2. Perhitungan parameter RFM berdasarkan customer_id
rfm_df = orders_items_merged.merge(customers_df, on="customer_id", how="inner")

rfm_df = rfm_df.groupby(by="customer_id", as_index=False).agg({
    "order_purchase_timestamp": "max", # Mengambil tanggal order terakhir
    "order_id": "nunique",             # Menghitung jumlah frekuensi order
    "price": "sum"                     # Menghitung total pengeluaran (monetary)
})

rfm_df.columns = ["customer_id", "max_order_timestamp", "frequency", "monetary"]

# 3. Perhitungan Recency (jumlah hari sejak transaksi terakhir terhadap tanggal transaksi paling akhir di dataset)
rfm_df["max_order_timestamp"] = pd.to_datetime(rfm_df["max_order_timestamp"]).dt.date
recent_date = pd.to_datetime(orders_df["order_purchase_timestamp"]).dt.date.max()
rfm_df["recency"] = rfm_df["max_order_timestamp"].apply(lambda x: (recent_date - x).days)

# Penghapusan kolom timestamp yang tidak diperlukan lagi
rfm_df.drop("max_order_timestamp", axis=1, inplace=True)

print("--- Hasil RFM Analysis ---")
print(rfm_df.head())
```

```python
# Visualisasi Pelanggan Terbaik Berdasarkan RFM Analysis

fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(30, 6))

colors = ["#72BCD4", "#72BCD4", "#72BCD4", "#72BCD4", "#72BCD4"]

# Grafik Recency (semakin kecil nilai recency, semakin baik/baru transaksi terakhirnya)
sns.barplot(y="recency", x="customer_id", data=rfm_df.sort_values(by="recency", ascending=True).head(5), hue="customer_id", palette=colors, legend=False, ax=ax[0])
ax[0].set_ylabel(None)
ax[0].set_xlabel(None)
ax[0].set_title("Best Customer By Recency (days)", loc="center", fontsize=18)
ax[0].tick_params(axis='x', labelsize=12, rotation=15)

# Grafik Frequency (semakin tinggi frequency, semakin sering bertransaksi)
sns.barplot(y="frequency", x="customer_id", data=rfm_df.sort_values(by="frequency", ascending=False).head(5), hue="customer_id", palette=colors, legend=False, ax=ax[1])
ax[1].set_ylabel(None)
ax[1].set_xlabel(None)
ax[1].set_title("Best Customer By Frequency", loc="center", fontsize=18)
ax[1].tick_params(axis='x', labelsize=12, rotation=15)

# Grafik Monetary (semakin tinggi monetary, semakin besar pengeluaran pelanggan)
sns.barplot(y="monetary", x="customer_id", data=rfm_df.sort_values(by="monetary", ascending=False).head(5), hue="customer_id", palette=colors, legend=False, ax=ax[2])
ax[2].set_ylabel(None)
ax[2].set_xlabel(None)
ax[2].set_title("Best Customer By Monetary", loc="center", fontsize=18)
ax[2].tick_params(axis='x', labelsize=12, rotation=15)

plt.suptitle("Best Customer Based on RFM Parameters", fontsize=22)
plt.show()
```

**Insight:** (Opsional)
- Seluruh customer_id pada sampel di atas memiliki nilai Frequency = 1. Hal ini menunjukkan bahwa sebagian besar pelanggan dalam segmen ini baru melakukan transaksi sekali saja dan belum menjadi pelanggan yang repeat order.
- Nilai recency tercatat sangat tinggi, berkisar antara 198 hingga 596 hari sejak transaksi terakhir mereka. Artinya, pelanggan-pelanggan ini sudah sangat lama tidak aktif berbelanja dan masuk dalam kategori pelanggan pasif atau berisiko hilang.
- Meskipun baru bertransaksi sekali, nilai monetary yang dikeluarkan cukup bervariasi (mulai dari ~54.90 hingga ~179.99). Ini menunjukkan bahwa beberapa pelanggan memiliki daya beli yang cukup baik meskipun pembeliannya belum berulang.
- Karena tingginya jumlah pelanggan dengan frequency 1 dan recency yang sudah lama, perusahaan perlu merancang strategi re-engagement campaign (seperti mengirimkan email pengingat, penawaran khusus, atau diskon win-back) untuk mendorong mereka agar kembali melakukan pembelian kedua.

### Geospatial Analysis

Penerapan Geospatial Analysis adalah untuk menganalisis sebaran pesanan, pendapatan (revenue), serta performa logistik berdasarkan wilayah geografis (state dan city) dari pelanggan.

```python
# Penggabungan orders, customers, dan order_items
geo_df = orders_df.merge(customers_df, on="customer_id", how="inner")
geo_df = geo_df.merge(order_items_df, on="order_id", how="inner")

# Agregasi data berdasarkan state (negara bagian)
state_analysis_df = geo_df.groupby("customer_state").agg({
    "order_id": "nunique",      # Jumlah pesanan unik
    "price": "sum",             # Total revenue/pendapatan
    "customer_id": "nunique"    # Jumlah pelanggan unik
}).reset_index()

state_analysis_df.columns = ["customer_state", "total_orders", "total_revenue", "total_customers"]
state_analysis_df.sort_values(by="total_revenue", ascending=False, inplace=True)

print("--- Ringkasan Analisis Geografis per State ---")
print(state_analysis_df.head(10))
```

```python
# Visualisasi Geospatial

# Visualisasi Total Revenue per Negara Bagian (State) khusus 10 besar
plt.figure(figsize=(12, 6))

top_state_df = state_analysis_df.head(10)
colors = ["#72BCD4" if state == "SP" else "#D3D3D3" for state in top_state_df["customer_state"]]

sns.barplot(
    x="total_revenue",
    y="customer_state",
    data=top_state_df,
    hue="customer_state",
    palette=colors,
    legend=False
)
plt.title("Top 10 Negara Bagian Berdasarkan Total Pendapatan (Revenue)", fontsize=16, pad=15)
plt.xlabel("Total Revenue", fontsize=12)
plt.ylabel("State (Negara Bagian)", fontsize=12)
plt.show()
```

**Insight:**
- Analisis spasial menunjukkan bahwa sebagian besar transaksi dan pendapatan perusahaan didominasi oleh negara bagian tertentu, terutama São Paulo (SP), yang mencatatkan volume pesanan dan revenue tertinggi dibandingkan wilayah lainnya.
- Wilayah luar São Paulo menunjukkan kontribusi yang lebih kecil, yang mengindikasikan adanya peluang atau tantangan logistik (seperti durasi pengiriman yang lebih lama) di luar pusat kota utama. Hal ini sangat relevan untuk merancang strategi pemasaran atau penambahan titik gudang baru di masa mendatang.
